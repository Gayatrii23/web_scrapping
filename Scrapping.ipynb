```python
import requests
from bs4 import BeautifulSoup

url = "https://raw.githubusercontent.com/Gayatrii23/award/main/oscar"
response = requests.get(url)
html_text = response.text

soup = BeautifulSoup(html_text, 'html.parser')
table = soup.find('table')

data = []
for index in table.find_all('tr'):
    cells = index.find_all(['th', 'td'])
    data.append([cell.get_text(strip=True) for cell in cells])

for row in data:
    print(row)

```

```python
l = []
for i in data[1:]:
    l.append(i[0].strip())

ll = []
for i in l:
    parts = i.split()
    ll.append(' '.join(parts[:2]))

print(ll)
print(len(ll))
```
